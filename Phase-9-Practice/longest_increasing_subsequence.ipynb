# Dynamic Programming: Longest Increasing Subsequence (LIS)

## Problem Statement

Given an array, find the **longest subsequence** where elements are in **strictly increasing order**.

A subsequence doesn't need to be contiguous.

### Example
```
Input: [10, 9, 2, 5, 3, 7, 101, 18]
Output: 4

Explanation: LIS is [2, 3, 7, 101], length = 4
```

---

## Approaches

### Brute Force: O(2^n)
- Generate all 2^n subsequences
- Check which ones are increasing
- Track longest

### DP: O(n²) or O(n log n)
- **dp[i]** = length of LIS ending at index i
- For each i, check all j < i where nums[j] < nums[i]
- dp[i] = 1 + max(dp[j]) if nums[j] < nums[i] else 1

---

## Approach 1: DP - O(n²)

````python
def lengthOfLIS_DP(nums):
    """
    Find LIS length using DP (O(n²))
    
    dp[i] = length of LIS ending at index i
    """
    if not nums:
        return 0
    
    n = len(nums)
    dp = [1] * n  # Each element is LIS of length 1
    
    for i in range(1, n):
        for j in range(i):
            # If nums[j] < nums[i], we can extend LIS ending at j
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    
    return max(dp)

# Test
nums = [10, 9, 2, 5, 3, 7, 101, 18]
result = lengthOfLIS_DP(nums)
print(f"Array: {nums}")
print(f"LIS length (DP O(n²)): {result}")
````

## Approach 2: Binary Search - O(n log n)

````python
import bisect

def lengthOfLIS_BinarySearch(nums):
    """
    Find LIS length using binary search (O(n log n))
    
    Maintain array tails where tails[i] = smallest ending value 
    of all increasing subsequences of length i+1
    """
    tails = []  # tails[i] = smallest tail of LIS with length i+1
    
    for num in nums:
        # Find position to insert num
        pos = bisect.bisect_left(tails, num)
        
        # If pos == len(tails), we have a new longest LIS
        if pos == len(tails):
            tails.append(num)
        else:
            # Replace the element at pos (to keep tail as small as possible)
            tails[pos] = num
    
    return len(tails)

# Test
nums = [10, 9, 2, 5, 3, 7, 101, 18]
result = lengthOfLIS_BinarySearch(nums)
print(f"Array: {nums}")
print(f"LIS length (Binary Search O(n log n)): {result}")
````

## Test Cases

````python
test_cases = [
    ([10, 9, 2, 5, 3, 7, 101, 18], 4, "[2,3,7,101] or [2,3,7,18]"),
    ([0, 1, 0, 4, 4, 4, 3, 4, 1], 2, "Any 2-length increasing sequence"),
    ([1, 2, 3, 4, 5], 5, "Already sorted"),
    ([5, 4, 3, 2, 1], 1, "Decreasing - all are LIS of length 1"),
    ([3, 10, 2, 1, 20], 3, "[3, 10, 20]"),
]

print("Comparing O(n²) vs O(n log n):")
print()

for i, (nums, expected, explanation) in enumerate(test_cases, 1):
    result_dp = lengthOfLIS_DP(nums)
    result_bs = lengthOfLIS_BinarySearch(nums)
    
    print(f"Test {i}: {nums}")
    print(f"  DP (O(n²)): {result_dp}")
    print(f"  Binary Search (O(n log n)): {result_bs}")
    print(f"  Expected: {expected}")
    print(f"  Example LIS: {explanation}")
    
    assert result_dp == expected, f"DP failed!"
    assert result_bs == expected, f"Binary search failed!"
    print(f"  ✅ Both correct!")
    print()
````

## Complexity Analysis

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Brute Force** | O(2^n) | O(1) | Simple | Too slow |
| **DP** | O(n²) | O(n) | Intuitive | Slower for large n |
| **Binary Search** | O(n log n) | O(n) | Optimal! | Trickier logic |

**For n = 10,000**:
- Brute force: 10^3000 operations (impossible!)
- DP: 100,000,000 operations (10 seconds)
- Binary search: 130,000 operations (instant!)

## How DP Works (Step by Step)

```
Array: [10, 9, 2, 5, 3, 7, 101, 18]

i=0: nums[0]=10, dp=[1]
i=1: nums[1]=9,  dp=[1, 1]        (9 < 10, so LIS is just [9])
i=2: nums[2]=2,  dp=[1, 1, 1]     (2 is smallest, LIS is just [2])
i=3: nums[3]=5,  dp=[1, 1, 1, 2]  (5 > 2, LIS is [2,5])
i=4: nums[4]=3,  dp=[1, 1, 1, 2, 2]  (3 > 2, LIS is [2,3])
i=5: nums[5]=7,  dp=[1, 1, 1, 2, 2, 3]  (7 > 5 and 7 > 3, LIS is [2,3,7] or [2,5,7])
i=6: nums[6]=101, dp=[1, 1, 1, 2, 2, 3, 4]  (101 > all, LIS is [2,3,7,101])
i=7: nums[7]=18, dp=[1, 1, 1, 2, 2, 3, 4, 4]  (18 < 101, but [2,3,7,18] is also length 4)

Answer: max(dp) = 4
```

## Interview Tips

✅ **First solution**: DP O(n²) - intuitive and correct

✅ **Optimization**: "Can we do better? Yes, with binary search O(n log n)"

✅ **Explain**: "We maintain smallest tail of each LIS length. For each new number, we find where it fits using binary search."

✅ **Tradeoff**: "DP is easier to explain, binary search is faster for large inputs"

✅ **Follow-up**: "Can you reconstruct the actual LIS (not just length)?" (Yes, with backtracking)
