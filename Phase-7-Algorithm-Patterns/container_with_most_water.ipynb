# Two Pointers: Container With Most Water (LeetCode 11)

## Problem Statement

Given an array `height[]` of vertical lines, find two lines that, together with the x-axis, form a container that holds the **maximum water**.

The water trapped = `min(height[i], height[j]) * (j - i)`

### Example
```
Input: height = [1,8,6,2,5,4,8,3,7]
Output: 49

Visual:
  |     |     |           <- lines at index 1 and 8
  |     | | |   | | |     <- water trapped = min(8,7) * (8-1) = 49
  | | | | | | | | | |
  -----------------------
```

---

## Brute Force vs Two Pointers

### ❌ Brute Force: O(n²)
- Try every pair (i, j)
- Calculate water = min(height[i], height[j]) * (j - i)
- Track maximum

### ✅ Two Pointers: O(n)
- Start with left=0, right=n-1 (widest container)
- Calculate water
- Move the pointer with **shorter height** inward
- Why? Only shorter line limits water. Moving the longer line won't help (still limited by shorter).

---

## Code Implementation

````python
def maxArea(height):
    """
    Find two lines that form maximum water container.
    
    Args:
        height: List of integers
    
    Returns:
        Maximum water trapped
    """
    left = 0
    right = len(height) - 1
    max_water = 0
    
    while left < right:
        # Calculate water between left and right
        width = right - left
        h = min(height[left], height[right])
        water = width * h
        max_water = max(max_water, water)
        
        # Move the pointer with shorter height
        # (moving the taller one won't help)
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    
    return max_water

# Test
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
result = maxArea(height)
print(f"Heights: {height}")
print(f"Max water: {result}")
````

## Test Cases

````python
# Test Case 1: Example
print("Test 1: Standard example")
h1 = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print(f"Heights: {h1}")
print(f"Max water: {maxArea(h1)}")
print(f"Expected: 49 (lines at index 1 and 8, heights 8 and 7)")
print()

# Test Case 2: Simple
print("Test 2: Simple case")
h2 = [2, 3, 4, 5, 18, 17, 6]
print(f"Heights: {h2}")
print(f"Max water: {maxArea(h2)}")
print()

# Test Case 3: Edge case (2 lines)
print("Test 3: Two lines")
h3 = [1, 1]
print(f"Heights: {h3}")
print(f"Max water: {maxArea(h3)}")
print(f"Expected: 1 (1 * 1 = 1)")
print()

# Test Case 4: Decreasing
print("Test 4: Decreasing heights")
h4 = [9, 8, 7, 6, 5, 4, 3, 2, 1]
print(f"Heights: {h4}")
print(f"Max water: {maxArea(h4)}")
print(f"Expected: 16 (first and second: min(9,8)*(1-0) = 8, but...)")
````

## Why Two Pointers Works

**Greedy Insight**: Moving the taller line inward will NEVER increase water.

**Proof**:
- Water = min(height[left], height[right]) × width
- If height[left] < height[right]:
  - Water is limited by height[left]
  - Moving right inward decreases width
  - Moving left inward might increase min (good!) or keep it same (bad)
  - So we must move left to have any chance
- If we move right inward instead:
  - Width decreases (always bad)
  - min stays same or decreases (bad or same)
  - Result: always worse or same

## Complexity Comparison

| Approach | Time | Space | Notes |
|----------|------|-------|-------|
| **Brute Force** | O(n²) | O(1) | Try every pair |
| **Two Pointers** | O(n) | O(1) | Optimal! |

**n = 1000**: Brute force = 1,000,000 operations. Two pointers = 1,000 operations. 1000x faster!
