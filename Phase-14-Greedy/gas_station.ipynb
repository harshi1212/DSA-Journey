# Greedy Algorithm: Gas Station Problem (LeetCode)

## Problem Statement

There are `n` gas stations on a circular route. At station `i`, you can get `gas[i]` liters. To go from station `i` to `i+1`, you need `cost[i]` liters.

Your tank has unlimited capacity. Find the **starting station** from which you can complete the circuit. Return -1 if impossible.

### Example
```
gas =  [1,2,3,4,5]
cost = [3,4,5,1,2]

Output: 3
Explanation: Start at station 3, get 4L, go to station 4 (need 1L, have 3L left),
get 5L more (have 8L), go to station 0 (need 2L, have 6L left), ...
```

---

## Greedy Insight

**Key Idea**: If total gas ≥ total cost, a solution always exists.

**Greedy Choice**: When we can't reach station `j` from station `i`, skip all stations between `i` and `j` and try starting from `j+1`.

**Why?** If we can't reach `j` from `i`, we can't reach it from any station between `i` and `j` either (because they had even less fuel).

---

## Code Implementation

````python
def canCompleteCircuit(gas, cost):
    """
    Find starting station to complete circuit.
    
    Args:
        gas: List of gas at each station
        cost: List of cost to next station
    
    Returns:
        Starting station index, or -1 if impossible
    """
    # Quick check: if total gas < total cost, impossible
    if sum(gas) < sum(cost):
        return -1
    
    start = 0
    tank = 0
    
    for i in range(len(gas)):
        tank += gas[i] - cost[i]
        
        # If tank goes negative, we can't reach next station
        # So try starting from next station
        if tank < 0:
            start = i + 1
            tank = 0
    
    return start

# Test
gas = [1, 2, 3, 4, 5]
cost = [3, 4, 5, 1, 2]
result = canCompleteCircuit(gas, cost)
print(f"Gas:  {gas}")
print(f"Cost: {cost}")
print(f"Start from station: {result}")
````

## Test Cases

````python
# Test Case 1: Example from problem
print("Test 1: Can complete circuit")
gas1 = [1, 2, 3, 4, 5]
cost1 = [3, 4, 5, 1, 2]
result1 = canCompleteCircuit(gas1, cost1)
print(f"Gas:  {gas1}")
print(f"Cost: {cost1}")
print(f"Start: {result1}")
print()

# Test Case 2: Impossible (more cost than gas)
print("Test 2: Impossible case")
gas2 = [2]
cost2 = [3]
result2 = canCompleteCircuit(gas2, cost2)
print(f"Gas:  {gas2}")
print(f"Cost: {cost2}")
print(f"Start: {result2}")
print()

# Test Case 3: Only one solution
print("Test 3: Only station 0 works")
gas3 = [5, 1, 2, 3, 4]
cost3 = [4, 4, 1, 5, 1]
result3 = canCompleteCircuit(gas3, cost3)
print(f"Gas:  {gas3}")
print(f"Cost: {cost3}")
print(f"Start: {result3}")
print()

# Verify the answer for Test 3
if result3 != -1:
    print(f"Verification for starting at station {result3}:")
    tank = 0
    for i in range(len(gas3)):
        idx = (result3 + i) % len(gas3)
        tank += gas3[idx]
        print(f"  Station {idx}: +{gas3[idx]}L (tank={tank}L), need {cost3[idx]}L", end="")
        tank -= cost3[idx]
        print(f" -> tank={tank}L")
        if tank < 0:
            print("  ❌ Failed!")
            break
    if tank >= 0:
        print("  ✅ Success!")
````

## Complexity Analysis

| Aspect | Complexity | Explanation |
|--------|-----------|-------------|
| **Time Complexity** | O(n) | Single pass through stations |
| **Space Complexity** | O(1) | Only using variables |

---

## Why Greedy Works

**Claim**: If total gas ≥ total cost, there exists a starting position.

**Proof**:
1. If we fail at station `i`, it means: `tank < cost[i]`
2. This means sum(gas[start...i-1]) < sum(cost[start...i-1])
3. Since total gas ≥ total cost, we have enough for the entire circuit
4. Any station before `i` would fail even sooner (less accumulated fuel)
5. So we try starting from `i+1`
6. Eventually we find a valid start

---

## Common Mistakes

❌ **Mistake 1**: Try every starting position (brute force O(n²))
- Greedy is O(n)! Skip unnecessary checks.

❌ **Mistake 2**: Forget to check if total gas < total cost
- Easy optimization to eliminate impossible cases

❌ **Mistake 3**: Use different logic for the circle
- Just reset `start = i+1` when tank goes negative
