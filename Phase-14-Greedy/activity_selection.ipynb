# Greedy Algorithm: Activity Selection Problem

## Problem Statement

You have `n` activities with start and end times. Each activity can only be done by one person. Find the **maximum number of non-overlapping activities** you can select.

### Real-World Analogy
Imagine you have a movie theater and 10 movie screenings today. Each screening has a start and end time. You can only watch one movie at a time. How many movies can you watch without them overlapping?

### Example
```
Activities: [(1,3), (2,4), (3,5), (4,6), (5,7)]
Output: 3 activities
Selected: [(1,3), (3,5), (5,7)]  or  [(1,3), (4,6)]  etc.
```

---

## Greedy Approach

**Key Insight**: Always pick the activity that **ends earliest**.

**Why?** By finishing early, you leave maximum room for future activities.

### Algorithm Steps
1. Sort activities by **end time**
2. Select the first activity (ends earliest)
3. Remove all overlapping activities
4. Repeat until no activities left

---

## Code Implementation

````python
def activity_selection(activities):
    """
    Find maximum number of non-overlapping activities.
    
    Args:
        activities: List of tuples (start, end)
    
    Returns:
        List of selected activities
    """
    # Sort by end time (greedy choice)
    activities.sort(key=lambda x: x[1])
    
    selected = []
    last_end = float('-inf')  # Track when last activity ended
    
    for start, end in activities:
        # If this activity starts after last one ended, select it
        if start >= last_end:
            selected.append((start, end))
            last_end = end
    
    return selected

# Test
activities = [(1,3), (2,4), (3,5), (4,6), (5,7)]
result = activity_selection(activities)
print(f"Input: {activities}")
print(f"Selected: {result}")
print(f"Count: {len(result)}")
````

## Test Cases

````python
# Test Case 1: Basic example
print("Test 1: Basic overlapping activities")
activities1 = [(1,3), (2,4), (3,5), (4,6), (5,7)]
result1 = activity_selection(activities1)
print(f"Activities: {activities1}")
print(f"Selected: {result1}")
print(f"Count: {len(result1)}")
print()

# Test Case 2: No overlaps
print("Test 2: No overlaps")
activities2 = [(1,2), (3,4), (5,6)]
result2 = activity_selection(activities2)
print(f"Activities: {activities2}")
print(f"Selected: {result2}")
print(f"Count: {len(result2)}")
print()

# Test Case 3: Single activity
print("Test 3: Single activity")
activities3 = [(1,5)]
result3 = activity_selection(activities3)
print(f"Activities: {activities3}")
print(f"Selected: {result3}")
print(f"Count: {len(result3)}")
print()

# Test Case 4: All overlapping
print("Test 4: All overlapping (only pick shortest)")
activities4 = [(1,10), (2,8), (3,5), (4,7)]
result4 = activity_selection(activities4)
print(f"Activities: {activities4}")
print(f"Selected: {result4}")
print(f"Count: {len(result4)}")
````

| **Space Complexity** | O(n) | Storing result list |

---

## Why This Greedy Works

**Greedy Choice Property**: Picking the activity that ends earliest is always part of an optimal solution.

**Why?** If we don't pick the earliest-ending activity, we waste time that could fit more activities.

**Proof Sketch**:
- Suppose optimal solution doesn't include activity with earliest end time
- Replace the first activity in optimal with earliest-ending activity
- This doesn't break any constraints (since it ends even earlier)
- Result is still optimal
- By induction, greedy is optimal

## Common Mistakes

❌ **Mistake 1**: Sorting by start time instead of end time
- Wrong! You want to leave room for future activities

❌ **Mistake 2**: Greedy doesn't work here
- This IS one of the classic greedy algorithm success stories!

❌ **Mistake 3**: Using `start > last_end` instead of `start >= last_end`
- Wrong! Activities that touch at endpoints don't overlap
